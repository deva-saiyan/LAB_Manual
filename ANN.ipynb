```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
data = {
    "PatientID": ["P1", "P2", "P3", "P4", "P5"],
    "Age": [25, 35, 45, 55, 65],
    "CoughSeverity": [2, 4, 6, 8, 9],
    "OxygenLevel": [98, 95, 92, 88, 84],
    "Disease": ["Healthy", "Healthy", "Mild", "Severe", "Severe"]
}

df = pd.DataFrame(data)
df
```

```python
label = LabelEncoder()
df['Disease'] = label.fit_transform(df['Disease'])
```

```python
x = df[["Age", "CoughSeverity", "OxygenLevel"]]
y = df["Disease"]
```

```python
x_train , x_test , y_train , y_test = train_test_split(
    x ,y , test_size=0.3 , random_state=28)
```

```python
scalar = StandardScaler()
x_train = scalar.fit_transform(x_train)
x_test = scalar.fit_transform(x_test)
```

```python
from sklearn.neural_network import MLPClassifier

ANN = MLPClassifier(
    hidden_layer_sizes=(5,),
    activation='relu',
    solver='adam',
    max_iter=1000,
    random_state=50
)
```

```python
ANN.fit(x_train, y_train)   

y_pred = ANN.predict(x_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
print(classification_report(
    y_test,
    y_pred,
    labels=[0, 1, 2],
    target_names=label.classes_
))
```

```python

test_data = pd.DataFrame({
    "Age": [28, 50, 60, 42],
    "CoughSeverity": [3, 7, 9, 5],
    "OxygenLevel": [97, 90, 85, 93]
})

print(test_data)
```

```python
test_scaled = scalar.transform(test_data)

y_pred_test = ANN.predict(test_scaled)
pred_labels = label.inverse_transform(y_pred_test)

test_data["PredictedDisease"] = pred_labels

print(test_data)
```
